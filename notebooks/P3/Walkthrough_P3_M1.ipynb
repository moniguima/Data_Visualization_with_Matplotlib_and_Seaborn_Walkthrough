#### **Function:** `DataFrame.rename`

The method **`DataFrame.rename()`** changes the labels of columns or rows in a DataFrame. Here it is used with a **function** instead of a dictionary, meaning it applies the same transformation to *all* column names.

---

##### **Code Example**

```python
df.rename(
    columns = lambda x: x.strip().lower().replace("-", "_"), 
    inplace = True
)
```

---

##### **Step-by-Step Explanation**

1. **`columns=`**

   * Tells Pandas you want to rename the **columns** (not the row index).

2. **`lambda x: ...`**

   * Instead of mapping old → new names manually, we pass a **function** (`lambda`) that is applied to each column name (`x`).
   * This is useful when you want to apply the same cleaning rule to all names.

3. **`x.strip()`**

   * Removes any leading or trailing spaces in the column name.
   * Example: `" Temp "` → `"Temp"`.

4. **`.lower()`**

   * Converts the cleaned name to lowercase.
   * Example: `"Temp"` → `"temp"`.

5. **`.replace("-", "_")`**

   * Replaces any hyphens (`-`) with underscores (`_`).
   * Example: `"temp-value"` → `"temp_value"`.

6. **`inplace=True`**

   * Applies the changes directly to the existing DataFrame `df`.
   * Without this, Pandas would return a *new* DataFrame instead.

---

##### **Result**

All column names in `df` will now:

* Have no extra spaces,
* Be lowercase,
* Use underscores instead of hyphens.

Example transformation:

```
" Temp-Value "   →   "temp_value"
```

---

**In short:**
This line standardizes column names in the DataFrame, making them clean, consistent, and easier to work with in code.

The **`DataFrame.sample()`** method returns a random subset of rows from a DataFrame. It’s often used for quick inspection, debugging, or creating smaller samples for testing or exploration.

---

##### **Code Example**

```python

```

---

##### **Result**

`df.sample(5)` returns 5 random rows from the DataFrame, with all their columns.

### 7. Replace some values in categorical columns in larger classes. 
This will reduce the number of unique values in these columns and help in creating more readable plots in later milestones.

In Step 3, when you apply the function unique() to the columns of DataFrame, you observe that:
- The column workclass has the unique values 'State-gov’, 'Self-emp-not-inc’, 'Private’, 'Federal-gov’, 'Local-gov’, 'Self-emp-inc’, ‘Without-pay’. Make the following replacements:
    - 'State-gov', 'Federal-gov', 'Local-gov' -> 'Gov' 
    - 'Self-emp-not-inc', 'Self-emp-inc' -> 'Self-Employed'
- The column education has the unique values 'Bachelors’, 'HS-grad’, '11th’, 'Masters’, '9th’, 'Some-college’, 'Assoc-acdm’, '7th-8th’, 'Doctorate’, 'Assoc-voc’, 'Prof-school’, '5th-6th’, '10th’, 'Preschool’, '12th’, ‘1st-4th’. Make the following replacements:
    - '11th’, '9th’, '7th-8th’, '5th-6th’, '10th’, '12th’, ‘1st-4th’ -> ‘School’
- The column marital-status has the unique values 'Never-married’, 'Married-civ-spouse’, 'Divorced’, 'Married-spouse-absent’, 'Separated’, 'Married-AF-spouse’, ‘Widowed’. Make the following replacements:
    - 'Married-civ-spouse’, 'Married-spouse-absent’, ‘Married-AF-spouse’ -> ‘Married’.
- The column occupation has the unique values 'Adm-clerical’, 'Exec-managerial’, 'Handlers-cleaners’, 'Prof-specialty’, 'Other-service’, 'Sales’, 'Transport-moving’, 'Farming-fishing’, 'Machine-op-inspct’, 'Tech-support’, 'Craft-repair’, 'Protective-serv’, 'Armed-Forces’, ‘Priv-house-serv’. Make the following replacements:
    - 'Adm-Clerical' -> 'Clerk', 
    - 'Exec-managerial' -> 'Manager', 
    - 'Prof-specialty' -> 'Professional', 
    - 'Handlers-cleaners', 'Other-service', 'Transport-moving', 'Machine-op-inspct', 'Tech-support', 'Craft-repair', 'Protective-serv', 'Prov-house-serv' -> 'Service', 
    - 'Farming-fishing' -> 'Farming'

````python
# define a dictionary of dictionaries to make the replacements for the columns
# workclass, education, marital_status, and occupation
super_class_dicts = {
    "workclass": {
        'State-gov': 'Gov',
        'Federal-gov': 'Gov',
        'Local-gov': 'Gov',
        'Self-emp-not-inc': 'Self-employed', 
        'Self-emp-inc': 'Self-employed' 
        },
    "education": {
        '11th': 'School', 
        '9th': 'School',
        '7th-8th': 'School', 
        '5th-6th': 'School', 
        '10th': 'School', 
        '12th': 'School', 
        '1st-4th': 'School'
    },
    "marital_status": {
        'Married-civ-spouse': 'Married', 
        'Married-spouse-absent': 'Married', 
        'Married-AF-spouse': 'Married'
    },
    "occupation":{
        'Adm-Clerical': 'Clerk',
        'Exec-managerial': 'Manager',
        'Prof-specialty': 'Professional',
        'Handlers-cleaners': 'Service',
        'Other-service': 'Service',
        'Transport-moving': 'Service', 
        'Machine-op-inspct': 'Service',
        'Tech-support': 'Service',
        'Craft-repair': 'Service', 
        'Protective-serv': 'Service', 
        'Prov-house-serv': 'Service',
        'Farming-fishing': 'Farming'
    }
}
````

````python
# Create new superclass columns
for col in ["workclass", "education","marital_status","occupation"]:
    col_sc = col+"_sc"
    dict_sc = super_class_dicts[col]
    df[col_sc] = df[col].apply(lambda x: x if x not in dict_sc else dict_sc[x])
    print(f"\nCreated column {col_sc}")
    print(df.groupby([col,col_sc]).size().to_frame(name="row_count").sort_index())
````

#### Code Explanation

##### 1. `super_class_dicts`

This is a dictionary of dictionaries.

* Each *key* is a column name (`workclass`, `education`, `marital_status`, `occupation`).
* Each *value* is another dictionary that maps fine-grained categories to broader "super classes".

Example:

```python
"workclass": {
    'State-gov': 'Gov',
    'Federal-gov': 'Gov',
    ...
}
```

So `"State-gov"` → `"Gov"`.

---

##### 2. The loop

```python
for col in ["workclass", "education","marital_status","occupation"]:
```

This iterates over the four categorical columns that need grouping.

---

##### 3. Create new column names

```python
col_sc = col+"_sc"
```

* For each column, it makes a new one with `_sc` (short for *super class*) appended.

  * `workclass` → `workclass_sc`
  * `education` → `education_sc`
  * etc.

---

##### 4. Get the right mapping dictionary

```python
dict_sc = super_class_dicts[col]
```

This extracts the dictionary of mappings for the current column.

---

##### 5. Apply transformation

```python
df[col_sc] = df[col].apply(lambda x: x if x not in dict_sc else dict_sc[x])
```

* For each row, look at the value `x` in the column.
* If `x` is a key in `dict_sc`, replace it with its mapped value.
* Otherwise, keep `x` unchanged.

Example:

* `df["workclass"] = "Federal-gov"` → `"Gov"`
* `df["workclass"] = "Private"` → stays `"Private"` (since not in dictionary).

---

##### 6. Print results

```python
print(f"\nCreated column {col_sc}")
```

* The counts (`943`, `2067`, …) show how many rows belong to each original category → super class.

Similar logic applies to `education`, `marital_status`, and `occupation`.

This code creates *simplified categories* (super classes) for multiple categorical columns in a DataFrame, by grouping detailed values (like `"Federal-gov"`, `"State-gov"`) into broader ones (like `"Gov"`). Then it shows how many rows fall into each original → super class mapping.
